```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# 模擬 ETF 的深層特徵向量 
# 特徵維度定義：[科技股權重, 金融股權重, 平均本益比(P/E), Beta值, 歷史年化波動度]
# 實務上這些資料可透過 yfinance 取得
etf_A_features = np.array([[0.45, 0.15, 25.4, 1.2, 0.18]])
etf_B_features = np.array([[0.42, 0.18, 24.8, 1.15, 0.17]]) # 與 A 性質相似
etf_C_features = np.array([[0.05, 0.40, 15.2, 0.8, 0.10]])  # 防禦型價值股 ETF，與 A 差異大

# 計算餘弦相似度 (Cosine Similarity)
similarity_AB = cosine_similarity(etf_A_features, etf_B_features)[0][0]
similarity_AC = cosine_similarity(etf_A_features, etf_C_features)[0][0]

print(f"ETF A 與 ETF B 的深層相似度: {similarity_AB * 100:.2f}% (適合做為替代推薦)")
print(f"ETF A 與 ETF C 的深層相似度: {similarity_AC * 100:.2f}% (不適合做為替代推薦)")
```
