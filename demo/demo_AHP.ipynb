```python
import numpy as np

def calculate_ahp_weights(matrix):
    n = matrix.shape[0]
    
    # 計算特徵值與特徵向量
    eigenvalues, eigenvectors = np.linalg.eig(matrix)
    
    # 找出最大特徵值及其對應的特徵向量
    max_index = np.argmax(np.real(eigenvalues))
    lambda_max = np.real(eigenvalues[max_index])
    principal_eigenvector = np.real(eigenvectors[:, max_index])
    
    # 將特徵向量正規化，使其總和為 1，即為所求的權重
    weights = principal_eigenvector / np.sum(principal_eigenvector)
    
    # 計算一致性比率 (CR)
    CI = (lambda_max - n) / (n - 1)
    # n=4 時的隨機指標 RI 為 0.90
    RI = 0.90 
    CR = CI / RI
    
    return weights, CR

# 模擬一位投資人的問卷填答結果 (建構 4x4 矩陣)
# 順序: [Cost, Liquidity, Diversification, Risk Fit]
A = np.array([
    [1,   3,   5,   2],   # Cost 比 Liquidity 重要 3 倍, 比 Div 重要 5 倍...
    [1/3, 1,   2,   1/2],
    [1/5, 1/2, 1,   1/3],
    [1/2, 2,   3,   1]
])

weights, CR = calculate_ahp_weights(A)

print("AHP 權重分配:")
print(f"Cost (成本): {weights[0]*100:.2f}%")
print(f"Liquidity (流動性): {weights[1]*100:.2f}%")
print(f"Diversification (分散化): {weights[2]*100:.2f}%")
print(f"Risk Fit (風險適配): {weights[3]*100:.2f}%")
print("-" * 30)
print(f"一致性比率 (CR): {CR:.4f}")

if CR < 0.1:
    print("邏輯一致性檢定通過，這些權重可以直接輸入至 AR-DEA 作為限制式。")
else:
    print("警告：CR 值大於 0.1，投資人填答存在邏輯矛盾，需重新評估。")
```
